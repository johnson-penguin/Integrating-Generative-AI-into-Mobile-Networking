```python
import os
import time
from langchain.document_loaders import UnstructuredMarkdownLoader
from langchain.text_splitter import RecursiveCharacterTextSplitter
from langchain.vectorstores import Chroma
from langchain_huggingface import HuggingFaceEmbeddings

# 設定資料與模型
markdown_folder = r"TSpec-LLM/Rel-16"
model_name = "sentence-transformers/all-MiniLM-L6-v2"
model_kwargs = {'device': 'cuda'}
embedding = HuggingFaceEmbeddings(model_name=model_name, model_kwargs=model_kwargs)

chunk_sizes = [300, 500, 700, 900, 1100]
chunk_overlap = 10

# 遞迴抓取所有 Markdown 檔案
markdown_files = []
for root, _, files in os.walk(markdown_folder):
    for file in files:
        if file.endswith(".md"):
            markdown_files.append(os.path.join(root, file))

print(f"📁 共發現 {len(markdown_files)} 個 Markdown 檔案")

start_time = time.time()
all_chunk_summary = {}

for chunk_size in chunk_sizes:
    print(f"\n🚀 開始處理 chunk_size = {chunk_size}")
    persist_root = f"3gpp_db_{chunk_size}"  # ✅ 根據 chunk_size 命名資料夾
    os.makedirs(persist_root, exist_ok=True)

    text_splitter = RecursiveCharacterTextSplitter(chunk_size=chunk_size, chunk_overlap=chunk_overlap)
    chunk_summary = {}

    for i, full_path in enumerate(markdown_files, 1):
        filename = os.path.basename(full_path)
        spec_name = os.path.splitext(filename)[0]

        print(f"\n🔹 處理第 {i}/{len(markdown_files)} 個：{spec_name}")

        loader = UnstructuredMarkdownLoader(full_path)
        documents = loader.load()

        if len(documents) == 0:
            print(f"⚠️  {spec_name} 沒有內容，跳過...\n")
            continue

        for doc in documents:
            doc.metadata["spec"] = spec_name
            doc.metadata["path"] = full_path

        splits = text_splitter.split_documents(documents)
        chunk_summary[spec_name] = len(splits)

        print(f"✅ 切割完成：{spec_name}，共 {len(splits)} 個 chunks")

        if len(splits) == 0:
            continue

        persist_path = os.path.join(persist_root, spec_name)
        os.makedirs(persist_path, exist_ok=True)

        Chroma.from_documents(
            documents=splits,
            embedding=embedding,
            persist_directory=persist_path
        ).persist()

        print(f"💾 已儲存至 {persist_path}")

    all_chunk_summary[chunk_size] = chunk_summary

end_time = time.time()

# 📊 統計輸出
print("\n📊 Chunk 統計總結：")
for chunk_size, summary in all_chunk_summary.items():
    print(f"\n🔸 chunk_size = {chunk_size}")
    total_chunks = sum(summary.values())
    for spec, count in summary.items():
        print(f" - {spec.ljust(20)}: {count} chunks")
    print(f"✅ 共產生 {total_chunks} chunks")

print(f"\n⏱️ 總處理時間：{end_time - start_time:.2f} 秒")
```
