```python
import papermill as pm
from datetime import datetime
from pathlib import Path

# 產生輸出資料夾
output_dir = Path("/home/aiml/johnson/testing_llm_5g_understand/result/notebook_runs")
output_dir.mkdir(parents=True, exist_ok=True)

# notebook 路徑清單
notebooks_to_run = [
    "/home/aiml/johnson/testing_llm_5g_understand/embedding_Experiment/llm_testing_teleqna_prompt_900.ipynb",
    "/home/aiml/johnson/testing_llm_5g_understand/embedding_Experiment/llm_testing_teleqna_prompt_1100.ipynb",
]

# 執行每一個 notebook 並存成帶時間戳的輸出
for nb_path in notebooks_to_run:
    timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
    name = Path(nb_path).stem.replace("(", "_").replace(")", "_")
    output_nb_path = output_dir / f"{name}_executed_{timestamp}.ipynb"
    
    print(f"🚀 Running: {nb_path}")
    pm.execute_notebook(
        input_path=nb_path,
        output_path=str(output_nb_path),
        parameters={},  # 若有需要注入參數可填這裡
    )
    print(f"✅ Finished: {output_nb_path}\n")
```
